# Weekly Sales Report

```python
#1. Creating Variables
Product_name = "Wireless Mouse"
Selling_price= 12500
Units_sold = 180
Sales_target = 200
Growth_rate = 15 
Current_stock = 250
Minimum_stock_level = 120
Product_available = True

#2. To calculate total weekly revenue
Revenue = Selling_price * Units_sold
print("Total Revenue:", Revenue)
```

```python
#3. To forecast next week's sales.
# Sales is expected to rise by 15%
Forecast_sales = Units_sold + (Units_sold * Growth_rate / 100)

print("Forecast Sales:", Forecast_sales)

#4. To forecast next week's revenue
Forecast_revenue = Forecast_sales * Selling_price

print("Forecast Revenue:", Forecast_revenue)

#5. To update the current stock
Updated_current_stock = Current_stock - Units_sold

print("Updated Current Stock:", Updated_stock)
```

```python
#6. To determine if the sales target(200) was achieved.
# The target is achieved if units sold is greater than or equal to
#the weekly sales target
print("Sales Target Achieved:", Units_sold >= Sales_target)

#7. To check if restocking is required
# Restock is required when the updated current stock is less than 
#the minimum stock level
print("Restocking Required:", Updated_current_stock < Minimum_stock_level)
```

```python
#8. To determine if the product is ready for promotion.
# The product is ready if it is available and the sales target was achieved

Product_available = True
Sales_target_achieved = False
print("Promotion Eligible:", Product_available and Sales_target_achieved) 

#9. To determine if demand is high.
# Demand is considered high if forecasted sales exceeds 200 unit OR
# forecasted revenue exceeds #2,500,000.
Forecast_sales = 207
Forecast_revenue = 2587500
High_demand = (Forecast_sales > 200 or Forecast_revenue > 2500000)
print("High Demand Forecast:", High_demand)
```

```python
# Generating the weekly sales report professionally
Report = f'''
=====================================================
       BRIGHTMART SALES FORECAST REPORT
======================================================

Product Name: {Product_name}

Selling Price: #{Selling_price}

Units Sold This Week: {Units_sold}

Weekly Revenue: #{Revenue}

Forecast Growth Rate: {Growth_rate}%

Forecast Sales: {Forecast_sales} Units

Forecast Revenue: #{Forecast_revenue}

Updated Current Stock: {Updated_current_stock} Units

Sales Target achieved: {Units_sold >= Sales_target}

Restocking Required: {Updated_current_stock < Minimum_stock_level}

Product Available: {Product_available}

Promotion Eligible: {Product_available and Sales_target_achieved}

High Demand Forecast: { High_demand}

========================================================'''

print(Report)
```
